```python
from sklearn import linear_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# 1. 切分数据（确保 X_train, y_train 存在）
X_train, X_test, y_train, y_test = train_test_split(
    diabetes_X, diabetes_y, test_size=0.2, random_state=42
)

# 2. 创建模型
model = linear_model.LinearRegression()

# 3. 关键步骤：先训练（fit），再预测（predict）
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# 4. 评估
mse = mean_squared_error(y_test, y_pred)
print("均方误差 (MSE):", mse)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, color="black", alpha=0.5, label="真实值")
plt.plot(X_test, y_pred, color="blue", linewidth=2, label="模型预测")
plt.xlabel("BMI (标准化后)")
plt.ylabel("疾病进展")
plt.title("我的第一个线性回归模型")
plt.legend()
plt.show()
```
